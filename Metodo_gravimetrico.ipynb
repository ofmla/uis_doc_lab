# 1. Método gravimétrico

## Introducción

En los levantamientos gravimétricos, la geología del subsuelo se investiga en función de las variaciones en el campo gravitacional de la Tierra causadas por las diferencias en la densidad de las rocas del subsuelo. Un concepto subyacente es la idea de un **cuerpo causante** - o fuente -, que es una unidad de roca con una densidad diferente a la que la rodea. Una fuente o cuerpo causante representa una zona subterránea de masa anómala que causa una perturbación localizada en el campo gravitacional conocida como gravedad anómala. Una amplia variedad de situaciones geológicas generan zonas de masa anómala. A pequeña escala, un relieve en la superficie de un sótano, como un valle profundo, puede generar anomalías mensurables. A mayor escala, las grandes anomalías de gravedad son generadas por plutones graníticos o cuencas sedimentarias. La interpretación de las anomalías de gravedad permite evaluar la posible profundidad y forma de la fuente.

## Teoria básica

La base del método gravimétrico es la **Ley de Gravitación de Newton**, que establece que la fuerza de atracción $F$ entre dos masas, $m_{1}$ y $m_{2}$, cuyas dimensiones son pequeñas con respecto a la distancia $r$ entre ellas, se obtiene mediante

\begin{equation}
  F=\dfrac{G\cdot m_{1}\cdot m_{2}}{r^{2}},
\end{equation}
 
donde $g$ es la constante gravitacional (6,67 $\times$ 10$^{-11}$ m$^{3}$kg$^{-1}$s$^{-2}$).

Considere el peso gravitacional de una Tierra esférica, estática y homogénea, con masa $M$ y radio $R$, sobre una pequeña masa $m$ en su superficie. Es relativamente sencillo demostrar que la masa de una esfera actúa como si estuviera concentrada en el centro de la esfera y, por sustitución en la ecuación anterior

\begin{equation}
F=\dfrac{G\cdot M}{R^{2}}m=mg
\end{equation}
 
La fuerza está relacionada con la masa a través de una aceleración, y el término $g=GM/R^{2}$ se conoce como **aceleración gravitacional**, o simplemente gravedad. El peso de la masa se expresa en mg.

En una Tierra así, la gravedad sería constante. Sin embargo, la forma elipsoidal, la rotación, el relieve irregular de la superficie y la distribución de masa interna de la Tierra hacen que la gravedad varíe a lo largo de su superficie.

El campo gravitacional se define de manera más útil en términos del *potencial gravitacional* $U$

\begin{equation}
U=\dfrac{G\cdot M}{r}
\end{equation}
 
Mientras que la aceleración gravitacional $g$ es una cantidad vectorial, que tiene una magnitud y una dirección y dirección (verticalmente hacia abajo), el potencial gravitacional 𝑈 es una cantidad escalar, que solo tiene magnitud. La primera derivada de $U$ en cualquier dirección da la componente de la gravedad en esa dirección. En consecuencia, el enfoque de campo potencial proporciona flexibilidad de cálculo.

## Unidades

¿Cuáles son las unidades de esta aceleración o $g$? La aceleración gravitacional en la superficie de la Tierra es de aproximadamente 9.8 m/s$^{2}$. Las variaciones de gravedad causadas por variaciones de densidad en el subsuelo son del orden de $\mu$m/s$^{2}$. Esta unidad de micrómetro por segundo se conoce como **unidad gravimétrica - gu**. En los levantamientos gravitacionales en tierra, se logra fácilmente una precisión de $\pm$0.1 gu, que corresponde a unas cien millonésimas del campo gravitacional normal. En el mar, la precisión alcanzada es considerablemente menor, alrededor de $\pm$10 gu.

La unidad de gravedad c.g.s es el **miligal** (1 gal= 1 cm/s$^{2}$; 1 mgal=10$^{-3}$ gal), equivalente a 10 gu. Normalmente, los geofísicos trabajan en unidades de miligales (0.001 gal). Algunos autores, especialmente en Europa, usan gu en lugar de gals. La aceleración gravitacional de la Tierra en estas unidades es 9.8 m/s$^{2}$ = 978 gal, o 978.031.85 mgal, o 9,780.318.5 gu en el ecuador.

## Anomalias de gravidade

Los gravímetros responden eficazmente solo a la componente vertical de la atracción gravitacional de una masa 
anómala. Considere el efecto gravitacional de una masa anómala $\delta g$, con componentes horizontal y vertical 
$\delta g_{x}$ y $\delta g_{z}$, respectivamente, sobre el campo de gravedad local $g$ y su representación en 
un diagrama vectorial.
<br />
<br />
<img style="float: center; width: 30%; height: 30%" src="body.svg">
<br />
<center>Relación entre el campo gravitacional y los componentes de la anomalía gravitatoria de una masa pequeña. </center> 

Resolviendo el rectángulo de fuerzas, tenemos

\begin{equation}
g+ \delta g=\sqrt{(g+\delta g_{z})^{2} + \delta g_{x}^{2}}=\sqrt{g^{2}+2g\delta g_{z} +\delta g_{z}^{2}+\delta g_{x}^{2}}
\end{equation} 
Los términos $\delta^{2}$ son insignificantemente pequeños y, por lo tanto, pueden ignorarse. La expansión binomial de 
la ecuación, entonces, da

\begin{equation}
g+\delta g \approx g+\delta g_{z}
\end{equation} 
de modo que

\begin{equation}
g+\delta \approx \delta g_{z}
\end{equation}
 
En consecuencia, las perturbaciones de gravedad medidas corresponden efectivamente a la componente vertical de 
la atracción del cuerpo causante.

## Anomalias gravimétricas de cuerpos de forma simple


Considere la atracción gravitacional de una masa puntual $m$ a una distancia $r$ de la masa. La atracción gravitacional $\Delta g_{r}$ en la dirección de la masa viene dada por

\begin{equation}
\Delta g_{r}=\dfrac{Gm}{r^{2}}
\end{equation}

Dado que solo se mide el componente vertical de la atracción $\Delta g_{z}$ causada por la masa, la anomalía gravitacional $\Delta g$ asociada con la masa es

\begin{equation}
\Delta g=\dfrac{Gm}{r^{2}}\cos \theta
\end{equation}

o 

\begin{equation}
\Delta g=\dfrac{Gmz}{r^{3}}
\end{equation}
 
Dado que una esfera actúa como si su masa estuviera concentrada en su centro, la ecuación anterior también corresponde a la anomalía gravitacional de una esfera cuyo centro está a una profundidad $z$

````python
from __future__ import unicode_literals
# Define `gsphere()`
def gsphere(a, x2, z2, del_density):
#calculate the gravity anomaly due to a sphere
#g-sphere = 4/3 rho G pi a^3 (1/(x^2+^2) (z)/(sqrt(x^2+z^2))
#distance units in meters, density contrast in kg /m3
#returns mgal

#a = radius of sphere;
#x2 is horizontal offset
#z2 is depth
#del_density is the density contrast  

    import math   # This will import math module

    #define consts.
    g=6.67e-11 #gravitational constant
    pi = 3.14159265358979

    grav=  4.0/3.0*pi*g*del_density*1.0e5*a*a*a*1/(x2*x2+z2*z2) *z2/(math.sqrt(x2*x2+z2*z2))
    return grav

#example driver for calculating gravity anomaly 
#due to buried sphere beneath a horizontal surface
def run():
    import matplotlib.pyplot as plt
    sphere_radius = 10. #m
    sphere_depth = 25. #m (sphere_depth > sphere_radius)
    sphere_density_contrast = 500. #kg/m3
    
    offset, gravity = [], []
    for i in range(0,79):
        offset.append(i * 2. -78.) #horizontal distance (m)
        gravity.append(gsphere(sphere_radius, offset[i], sphere_depth, sphere_density_contrast))
        #print ("offset: ", offset[i], gravity[i])

    plt.plot(offset, gravity,'bo',label='analítica')
    plt.ylabel('$g_{z}$ [mgal]')
    plt.xlabel('$Coordenada-x$ [m]')
    plt.grid(True)
    plt.legend(loc='upper left')
    plt.title('Anomalia gravimetrica de una masa puntual o esfera')
    plt.show()
    
run()
````

## Ambigüedad

Derivamos la atracción gravitacional asociada con un cuerpo esférico simple. La componente vertical de esta atracción resultó ser igual a:

\begin{equation}
\Delta g=\dfrac{4\pi\Delta \rho R^{3}Gz}{3(x^{2}+z^{2})^{3/2}}
\end{equation}

Tenga en cuenta que nuestra expresión para la aceleración gravitacional en una esfera contiene un término que 
describe los parámetros físicos del cuerpo esférico; su radio, $R$, y su contraste de densidad, $\Delta \rho$, en 
la forma

\begin{equation}
R^{3}\Delta \rho
\end{equation}
 
$R$ y $\Delta \rho$ son dos de los parámetros que describen la esfera que nos gustaría poder determinar a partir de nuestras observaciones gravitacionales (el tercero es la profundidad del centro de la esfera $z$). Es decir, nos gustaría calcular las aceleraciones gravitacionales pronosticadas dadas las estimaciones de $R$ y $\Delta \rho$, comparar con las observadas y luego variar $R$ y $\Delta \rho$ hasta que la aceleración pronosticada coincida con la aceleración observada.

Esto suena bastante simple, pero hay un problema importante: ¡hay un número infinito de combinaciones de $R$ y $\Delta \rho$ que producen exactamente la misma aceleración gravitacional! Por ejemplo, supongamos que encontramos valores para $R$ y $\Delta \rho$ que se ajustan a nuestras observaciones de tal manera que

\begin{equation}
R^{3}\Delta \rho=31,25
\end{equation}
 
Cualquier otra combinación de valores para $R$ y $\Delta \rho$ también se ajustará a las observaciones, siempre que $R$ al cubo por $\Delta \rho$ sea igual a 31,25.

Utilice la función `gsphere` para calcular los perfiles de respuesta de cuatro esferas de manera que $R^{3}\Delta \rho=$31,25. Nuestra incapacidad para resolver de manera única los parámetros que describen un modelo de la Tierra a partir de observaciones geofísicas no es exclusiva del método de gravedad, sino que está presente en todos los métodos geofísicos. Existe una variedad de expresiones para referirse a este problema: indeterminación del modelo, equivalencia del modelo y no unicidad, por nombrar algunas. No importa cómo se llame, siempre significa lo mismo; un método geofísico dado no puede definir de manera única la estructura geológica subyacente al levantamiento. Otra forma de pensar sobre este problema es darse cuenta de que un modelo de una estructura geológica puede definir de forma única el campo gravitacional sobre la estructura. El campo gravitacional, sin embargo, no puede definir exclusivamente la estructura geológica que lo produjo.

Si este es el caso, **¿cómo determinamos qué modelo es el correcto?** Para ello, debemos incorporar observaciones adicionales en las que basar nuestra interpretación. Estas observaciones adicionales presumiblemente limitarán el rango de modelos aceptables que debemos considerar al interpretar nuestras observaciones gravitacionales. Estas observaciones pueden incluir observaciones geológicas u observaciones de diferentes tipos de estudios geofísicos.

````python
from __future__ import unicode_literals

import math
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Gravity anomaly of a buried sphere
def gsphere(a, x2, z2, del_density):
    """
    a            : sphere radius (m)
    x2           : horizontal offset (m)
    z2           : depth to center (m)
    del_density  : density contrast (kg/m³)

    Returns gravity anomaly in mGal
    """
    g = 6.67e-11
    pi = math.pi

    grav = (
        4.0 / 3.0
        * pi
        * g
        * del_density
        * 1.0e5
        * a**3
        * z2
        / ((x2**2 + z2**2) ** 1.5)
    )

    return grav


# Interactive plotting function
def plot_sphere(radius=10, density=500, depth=25):

    offsets = np.arange(-78, 80, 2)

    gravity = [
        gsphere(radius, x, depth, density)
        for x in offsets
    ]

    plt.figure(figsize=(8, 5))
    plt.plot(offsets, gravity, 'bo-', label='Sphere anomaly')
    plt.xlabel('X coordinate (m)')
    plt.ylabel(r'$g_z$ (mGal)')
    plt.title('Gravity Anomaly of a Buried Sphere')
    plt.grid(True)
    plt.legend()
    plt.show()


# Sliders
radius_slider = widgets.FloatSlider(
    value=10,
    min=1,
    max=30,
    step=1,
    description='Radius (m)',
    continuous_update=False
)

density_slider = widgets.FloatSlider(
    value=500,
    min=-1000,
    max=1000,
    step=50,
    description='Density',
    continuous_update=False
)

depth_slider = widgets.FloatSlider(
    value=25,
    min=5,
    max=100,
    step=1,
    description='Depth (m)',
    continuous_update=False
)

# Interactive widget
interactive_plot = widgets.interactive(
    plot_sphere,
    radius=radius_slider,
    density=density_slider,
    depth=depth_slider
)

display(interactive_plot)
````

````python
%matplotlib notebook

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
import math

def gsphere(a, x2, z2, rho):
    g = 6.67e-11
    return (4/3)*math.pi*g*rho*1e5*a**3*z2/((x2**2+z2**2)**1.5)

depth = 25
radius0 = 10
rho0 = 500

x = np.arange(-78, 80, 2)

fig, ax = plt.subplots(figsize=(8,5))
plt.subplots_adjust(bottom=0.25)

y = [gsphere(radius0, xx, depth, rho0) for xx in x]
(line,) = ax.plot(x, y)

ax.set_xlabel("Offset (m)")
ax.set_ylabel("gz (mGal)")
ax.grid()

ax_radius = plt.axes([0.15, 0.10, 0.70, 0.03])
ax_rho = plt.axes([0.15, 0.05, 0.70, 0.03])

s_radius = Slider(ax_radius, "Radius", 1, 30, valinit=radius0)
s_rho = Slider(ax_rho, "Density", -1000, 1000, valinit=rho0)

def update(val):
    r = s_radius.val
    rho = s_rho.val

    y = [gsphere(r, xx, depth, rho) for xx in x]
    line.set_ydata(y)

    ax.relim()
    ax.autoscale_view()
    fig.canvas.draw_idle()

s_radius.on_changed(update)
s_rho.on_changed(update)

plt.show()
````
